```python
import pandas as pd
```

```python
orders = pd.read_csv("olist_orders_dataset.csv")
orders.head()
```

```python
orders.shape
```

```python
orders.info()
```

```python
orders["order_status"]
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])

orders.info()
```

```python
orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"] - orders["order_estimated_delivery_date"]
).dt.days

orders[
    [
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "delivery_delay_days",
    ]
].head(10)
```

```python
order_items = pd.read_csv("olist_order_items_dataset.csv")
order_items.head()
```

```python
orders_items = orders.merge(order_items, on="order_id", how="left")

print("orders shape:", orders.shape)
print("order_items shape:", order_items.shape)
print("merged shape:", orders_items.shape)

orders_items.head()
```

```python
orders_items = orders.merge(order_items, on="order_id", how="left")

print("orders shape:", orders.shape)
print("order_items shape:", order_items.shape)
print("merged shape:", orders_items.shape)

orders_items.head()
```

```python
order_payments = pd.read_csv("olist_order_payments_dataset.csv")
order_reviews = pd.read_csv("olist_order_reviews_dataset.csv")

print("order_payments shape:", order_payments.shape)
print("order_reviews shape:", order_reviews.shape)
```

```python
payments_agg = (
    order_payments.groupby("order_id")
    .agg(
        total_payment_value=("payment_value", "sum"),
        payment_count=("payment_value", "count"),
    )
    .reset_index()
)

print("payments_agg shape:", payments_agg.shape)
payments_agg.head()
```

```python
reviews_agg = (
    order_reviews.groupby("order_id")
    .agg(
        avg_review_score=("review_score", "mean"),
        review_count=("review_score", "count"),
    )
    .reset_index()
)

print("reviews_agg shape:", reviews_agg.shape)
reviews_agg.head()
```

```python
full_data = orders_items.merge(payments_agg, on="order_id", how="left")
full_data = full_data.merge(reviews_agg, on="order_id", how="left")

print("orders_items shape:", orders_items.shape)
print("full_data shape:", full_data.shape)

full_data.head()
```

```python
products = pd.read_csv("olist_products_dataset.csv")
category_translation = pd.read_csv("product_category_name_translation.csv")

products = products.merge(category_translation, on="product_category_name", how="left")

print("products shape:", products.shape)
products[
    ["product_id", "product_category_name", "product_category_name_english"]
].head()
```

```python
full_data = full_data.merge(
    products[["product_id", "product_category_name_english"]],
    on="product_id",
    how="left",
)

print("full_data shape:", full_data.shape)
full_data[["order_id", "product_id", "product_category_name_english"]].head()
```

```python
full_data.isnull().sum()
```

```python
missing_category = full_data[full_data["product_category_name_english"].isnull()]
print("Rows with missing category:", missing_category.shape[0])
print(
    "Of these, rows with a valid product_id:",
    missing_category["product_id"].notnull().sum(),
)
```

```python
problem_product_ids = missing_category.loc[
    missing_category["product_id"].notnull(), "product_id"
].unique()

sample_products = products[products["product_id"].isin(problem_product_ids)]
print(
    sample_products[
        ["product_id", "product_category_name", "product_category_name_english"]
    ].head(10)
)
print(
    "\nHow many of these have a blank product_category_name in the original products table:"
)
print(sample_products["product_category_name"].isnull().sum())
```

```python
category_fix = {
    "pc_gamer": "pc_gamer",
    "portateis_cozinha_e_preparadores_de_alimentos": "kitchen_portables_and_food_preparers",
}

full_data["product_category_name_english"] = full_data[
    "product_category_name_english"
].fillna(full_data["product_category_name"].map(category_fix))

full_data["product_category_name_english"] = full_data[
    "product_category_name_english"
].fillna("unknown")

print(full_data["product_category_name_english"].isnull().sum())
```

```python
product_to_category = products.set_index("product_id")["product_category_name"]

fallback_category = full_data["product_id"].map(product_to_category).map(category_fix)

full_data["product_category_name_english"] = full_data[
    "product_category_name_english"
].fillna(fallback_category)
full_data["product_category_name_english"] = full_data[
    "product_category_name_english"
].fillna("unknown")

print(full_data["product_category_name_english"].isnull().sum())
```

```python
customers = pd.read_csv("olist_customers_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")

full_data = full_data.merge(
    customers[["customer_id", "customer_city", "customer_state"]],
    on="customer_id",
    how="left",
)

full_data = full_data.merge(
    sellers[["seller_id", "seller_city", "seller_state"]], on="seller_id", how="left"
)

print("full_data shape:", full_data.shape)
full_data[["customer_id", "customer_state", "seller_id", "seller_state"]].head()
```

```python
full_data.to_csv("olist_full_data_cleaned.csv", index=False)
print("Saved! Final shape:", full_data.shape)
```
